```python
import math

class Triangle:
    def __init__(self, leg1, leg2):
        self.leg1 = leg1
        self.leg2 = leg2

    def get_info(self):
        return f"Прямоугольный треугольник со сторонами: {self.leg1} и {self.leg2}"

    def get_tan(self):
        if self.leg1 == 0:
            return "Ошибка: первый катет равен нулю"
        return self.leg2 / self.leg1

leg1 = float(input("Введите первый катет: "))
leg2 = float(input("Введите второй катет: "))

tr = Triangle(leg1, leg2)

print(tr.get_info())
print(f"Тангенс угла: {tr.get_tan()}")
```

```python
import math

class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def get_info(self):
        sign = "+" if self.imag >= 0 else "-"
        return f"Комплексное число: {self.real} {sign} {abs(self.imag)}i"

    def multiply_by_number(self, factor):
        new_real = self.real * factor
        new_imag = self.imag * factor
        return f"Результат умножения на {factor}: {new_real} + {new_imag}i"

    def get_argument_degrees(self):
        argument_radians = math.atan2(self.imag, self.real)
        return math.degrees(argument_radians)

real_part = float(input("Введите действительную часть: "))
imag_part = float(input("Введите мнимую часть: "))

c_num = ComplexNumber(real_part, imag_part)
print(c_num.get_info())

multiplier = float(input("Введите число для умножения: "))
print(c_num.multiply_by_number(multiplier))
print(f"Аргумент комплексного числа в градусах: {c_num.get_argument_degrees()}")
```

```python
import math

class Triangle:
    def __init__(self, leg1, leg2):
        self.leg1 = leg1
        self.leg2 = leg2

    def get_info(self):
        return f"Катеты: {self.leg1}, {self.leg2}"

    def get_alpha(self):
        return math.atan2(self.leg2, self.leg1)

class DetailedTriangle(Triangle):
    def __init__(self, leg1, leg2, beta):
        super().__init__(leg1, leg2)
        self.beta = beta

    def get_info(self):
        return f"{super().get_info()}, beta: {self.beta}"

    def get_diff(self):
        return self.beta - self.get_alpha()

leg1 = float(input())
leg2 = float(input())
beta = float(input())

t1 = Triangle(leg1, leg2)
print(t1.get_info())

t2 = DetailedTriangle(leg1, leg2, beta)
print(t2.get_info())
print(t2.get_diff())
```

```python
import cmath

class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def get_info(self):
        return f"Число: {self.real} + {self.imag}j"

class Circuit(ComplexNumber):
    def __init__(self, r, x_l):
        super().__init__(r, x_l)

    def get_info(self):
        return f"Сопротивление Z = {self.real} + {self.imag}j"

    def get_admittance(self):
        z = complex(self.real, self.imag)
        y = 1 / z
        return abs(y), cmath.phase(y)

r = float(input())
xl = float(input())

c = Circuit(r, xl)
print(c.get_info())

mod, arg = c.get_admittance()
print(f"Модуль: {mod}, Аргумент: {arg}")
```

```python
class Diamond:
    def __init__(self, name, weight, cut_quality):
        self.name = name
        self.weight = weight
        self.cut_quality = cut_quality

    def get_q(self):
        return 0.4 * self.weight + 0.6 * self.cut_quality

    def get_info(self):
        return f"Алмаз: {self.name}, вес: {self.weight}, огранка: {self.cut_quality}, Q: {self.get_q()}"

class ColoredDiamond(Diamond):
    def __init__(self, name, weight, cut_quality, color):
        super().__init__(name, weight, cut_quality)
        self.color = color

    def get_q(self):
        q = super().get_q()
        if self.color == "голубой":
            return q + 1
        elif self.color == "желтый":
            return q - 0.5
        return q

    def get_info(self):
        return f"Алмаз: {self.name}, вес: {self.weight}, огранка: {self.cut_quality}, цвет: {self.color}, Qp: {self.get_q()}"

d1 = Diamond("Кохинур", 105.6, 9)
print(d1.get_info())

d2 = ColoredDiamond("Хоуп", 45.52, 10, "голубой")
print(d2.get_info())
```
